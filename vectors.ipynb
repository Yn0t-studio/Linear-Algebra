```python
# plot a vector 

u = np.array([1, 2])
v = np.array([1, -2])

plt.figure(figsize=(8, 6))
plt.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1)
# Move vector v to the tip of the other vector
plt.quiver(u[0], u[1], v[0], v[1], angles='xy', scale_units='xy', scale=1, color='orange')
plt.quiver(v[0], v[1], u[0], u[1], angles='xy', scale_units='xy', scale=1, color='orange')


addition = u + v
plt.quiver(0, 0, addition[0], addition[1], angles='xy', scale_units='xy', scale=1, color='r')

# Reverse the direction of vector v and move it to the tip of vector u
plt.quiver(u[0], u[1], -v[0], -v[1], angles='xy', scale_units='xy', scale=1, color='orange')
subtraction = u - v
plt.quiver(0, 0, subtraction[0], subtraction[1], angles='xy', scale_units='xy', scale=1, color='g')

plt.xlim(-10, 10)
plt.ylim(-10, 10)
plt.grid()
plt.show()
```

The norm of a vector

$$\|v\| = \sqrt{v_1^2 + v_2^2 + \dots + v_n^2}$$

```python
v = np.array([4, 7])

# Compute the norm of the vector using the formula ||v|| = sqrt(v1^2 + v2^2 + ... + vn^2)
def norm(v):
    norm = 0
    for i in range(len(v)):
        norm += v[i]**2
    return math.sqrt(norm)

#
print(np.linalg.norm(v))
print(norm(v))
```

```python
v = np.array([6., 8.])
norm = np.linalg.norm(v)
print(norm)

for i in range(len(v)):
    v[i] = v[i] / norm
print(v)
print(np.linalg.norm(v))
```

```python
def set_vector_magnitude(vector, magnitude):
    """
    Set the magnitude of a vector to a given value.
    """
    normalized_vector = vector / np.linalg.norm(vector)
    return normalized_vector * magnitude

v = np.array([2, 4])
print("Original vector:", v)
print("Original magnitude:", np.linalg.norm(v))
v = set_vector_magnitude(v, 10)
print("New vector:", v)
print("New magnitude:", np.linalg.norm(v))
```
